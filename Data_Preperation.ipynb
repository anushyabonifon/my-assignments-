DATA PREPERATION

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, RobustScaler
data = {
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Female'],
    'City': ['Mumbai', 'Delhi', 'Chennai', 'Mumbai', 'Delhi'],
    'Size': ['Small', 'Large', 'Medium', 'Large', 'Small'],
    'Age': [25, 45, 32, 60, 28],
    'Fare': [15, 300, 85, 450, 20]
}
df = pd.DataFrame(data)
#Task 1:Apply the correct encoding method to each categorical column:
#Gender → Label Encoding
#City → One-Hot Encoding (drop one column to avoid the dummy variable trap)
#Size → Label Encoding with the correct order: Small=0, Medium=1, Large=2
le=LabelEncoder()
df['Gender']=le.fit_transform(df['Gender'])
size_mapping = {
    'Small': 0,
    'Medium': 1,
    'Large': 2
}
df['Size'] = df['Size'].map(size_mapping)
df = pd.get_dummies(df, columns=['City'], drop_first=True)
#Task 2:
#RobustScaler was chosen because the Fare column contains a potential outlier (450).
#RobustScaler uses the median and interquartile range (IQR), making it less sensitive to extreme values and therefore more appropriate than StandardScaler for this dataset."
scalar=RobustScaler()
df[['Age','Fare']]=scalar.fit_transform(df[['Age','Fare']])
print(df)
```
